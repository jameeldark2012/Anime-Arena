```python
import pandas as pd
```

```python
data = pd.read_csv("data/characters.csv")
```

```python
data.head()
```

```python
data = data.rename(columns={"mal_id": "character_id"})
data = data.rename(columns={"name": "character_name"})
data
```

```python
cols_to_drop = [
    "favorites",
    "about",
    "episodes",
]

data = data.drop(columns=[c for c in cols_to_drop if c in data.columns])
data
```

```python
idx = data.columns.get_loc("character_name") + 1
data.insert(idx, "claimed_by", -1)
data
```

```python
idx = data.columns.get_loc("character_name") + 1
data.insert(idx, "anime_id", -1)
data
```

```python
null_count = data["url"].isnull().sum()
null_count
```

```python
excess_duplicates = data[data.duplicated(subset=["character_id"])]
excess_duplicates[["character_id", "character_name"]]
```

```python
print(data.dtypes)
```

```python
data.to_csv("data/characters_cleaned.csv", index=False)
```
