```python
import pandas as pd
```

```python
data = pd.read_csv("../data/anime.csv")
```

```python
data.head()
```

```python
data = data.rename(columns={"mal_id": "anime_id"})
data = data.rename(columns={"title": "anime_name"})

data
```

```python
cols_to_drop = [
    "scored_by",
    "status",
    "episodes",
    "aired_from",
    "aired_to",
    "source",
    "producers",
    "licensors",
    "title_synonyms",
    "type",
    "members",
    "favorites",
    "duration",
    "rating",
    "nsfw",
    "themes",
    "demographics",
    "studios",
    "synopsis",
    "background",
    "pending_approval",
    "premiered_season",
    "premiered_year",
    "broadcast_day",
    "broadcast_time"
    
    
    
    
]
data = data.drop(columns=[c for c in cols_to_drop if c in data.columns])
data
```

```python
null_count = data["url"].isnull().sum()
null_count
```

```python
print(data.dtypes)
```

```python
excess_duplicates = data[data.duplicated(subset=["anime_name"])]

excess_duplicates[["anime_id", "anime_name"]]
```

```python
data = data.drop_duplicates(subset=["anime_name"])

data.shape
```

```python
data.to_csv("../data/anime_cleaned.csv", index=False)
```
